# Visualizing continuous to continuous relationships - Scatter plots

## Import Modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

## Read Data

We will use the Penguins data set for this example.

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Continuous to Continuous relationships

The fundamental plot for showing the relationship between 2 continuous variables is the **SCATTER PLOT**.

In Seaborn, the FIGURE LEVEL function for showing RELAIONSHIPS is the `sns.relplot()` function.

The `x` axis is associated with a continuous variable and the `y` axis is associated with a second continuous variable.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

```python
sns.relplot(data = penguins, x='bill_depth_mm', y='flipper_length_mm')

plt.show()
```

```python
penguins.info()
```

```python
sns.relplot(data = penguins, x='body_mass_g', y='bill_length_mm')

plt.show()
```

The underlying AXIS level function called by the FIGURE LEVEL `sns.relplot()` is the `sns.scatterplot()` function.

```python
fig, ax = plt.subplots()

sns.scatterplot(data = penguins, x='flipper_length_mm', y='bill_length_mm', ax=ax)

plt.show()
```

In Summary, the SCATTER PLOT shows the RELATIONSHIP between 2 continuous variables.

However, SCATTER PLOTS can show MORE THAN just 2 variables!!!!!

A very common practice is to INCLUDE a categorical variable via color!!! We can associate a categorical variable with the `hue` argument!!!!

Let's return to our first plot.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

But we have categorical variables, object data types, in the `penguins` data set!

```python
penguins.info()
```

What if we GROUP BY a categorical variable? In other words...we can study if the RELATIONSHIP between the 2 variables CHANGES across CATEGORIES!!!!

Let's assign the `species` variable to the `hue` argument in `sns.relplot()`.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

Instead of coloring by the categorical variable we can also use FACETS to SEPARATE the groups!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', col='species')

plt.show()
```

You can combine `hue` with `col` facets to get facets and colors.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species', col='species')

plt.show()
```

However...if you do NOT color by the same variable as the FACET...that means you can use DIFFERENT categorical variables for color and facet!!!

You can GROUP the relationship on 2 separate and distinct categorical variables! You are grouping by the COMBINATION of 2 categorical variables!

The combination on `species` and `island` on the relationship!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species', col='island')

plt.show()
```

We can add yet another categorical variable throw ROW facets!!!!

```python
penguins.info()
```

If we use ROW and COL facets and associate `hue` with a categorical variable we are effectively GROUPING the relationship between 2 continuous variables on the combination of 3 categorical variables!!!

```python
sns.relplot( data = penguins, x='flipper_length_mm', y='bill_depth_mm',
             hue='sex',
             col='island',
             row='species')

plt.show()
```

Remember to always check the COMBINATIONS between categoricals BEFORE looking at the relationships between continuous variables!!

```python
sns.catplot(data = penguins, x='island', hue='species', kind='count')

plt.show()
```

# Visualize continuous to continuous relationships via TREND PLOTS

The TREND PLOT is an extension of the scatter plot that adds further "quantitative" information about the relationship between two continuous variables.

## Import Modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

## Read data

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Scatter plot

Show the relationship between two continuous variables.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

We saw that by GROUPING or CONDITIONING on a categorical variable...our interpetation of the scatter plot changed!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

## Trend plots

We can add TREND LINES to visualize the **LINEAR** relationship between the two continuous variables.

TREND LINES correspond to the **BEST FIT LINE** between the two variables.

Later on...we will learn HOW this is line is created...but for now let's visualize it and let the line or TREND HELP US interpret the relationship!

The TREND is NOT created with the `sns.relplot()` function.

Instead, the TREND is created with the FIGURE LEVEL function `sns.lmplot()`.

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

We are viewing an ANTI or NEGATIVE relationship between the two variables!!!!

BUT...what happens if we GROUP or CONDITION ON a categorical variable?!??!?

Let's COLOR via the `hue` argument BY `species`.

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

GIVEN or GROUPING BY or CONDITIONING ON the `species` CHANGES our opinion about the RELATIONSHIP betwen the 2 continuous variables! GROUPING BY `species` reveals a POSITIVE relationship between the two!!!

We can of course create the TREND PLOT between 2 different continuous variabels!

```python
penguins.info()
```

```python
sns.lmplot(data = penguins, x='body_mass_g', y='bill_length_mm')

plt.show()
```

```python
sns.lmplot(data = penguins, x='body_mass_g', y='bill_length_mm', hue='species')

plt.show()
```

As with the scatter plot we can also GROUP BY or CONDITION ON a SECONDARY VARIABLE via FACETS!!!!!

Let's facet by the penguin `sex`.

```python
sns.lmplot(data = penguins, x='body_mass_g', y='bill_length_mm', hue='species', col='sex')

plt.show()
```

```python
sns.lmplot(data = penguins, x='body_mass_g', y='bill_length_mm', hue='sex', col='species')

plt.show()
```

A THIRD categorical variable can be included via ROW facets!

```python
sns.lmplot(data = penguins, x='body_mass_g', y='bill_length_mm', hue='sex', col='species', row='island')

plt.show()
```

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='sex', col='species', row='island')

plt.show()
```

The TREND PLOT does NOT connect the DOTS!!! The TREND PLOT is actual a MODEL and thus a MODEL is fit behind the scenes!

# Visualizing continuous to continuous relationships via CORRELATION PLOTS

This video is another extension of the SCATTER PLOT. But instead of looking at the RAW data, we will focus on a SINGLE SUMMARY STATISTIC that QUANTIFIES the linear relationship between the two continuous variables!!!!!

## Import Modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

## Read data

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Scatter plot

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

## Trend Plot

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

## Correlation coefficient

The correlation coefficient is between -1 and +1. A value of -1 means there is a perfect ANTI or NEGATIVE **linear** relationship between the two variables. A value of +1 means there is a perfect POSITIVE **linear** relationship between the two variables.

The correlation MATRIX can be calculated via the Pandas `.corr()` method.

```python
penguins.loc[ :, ['flipper_length_mm', 'bill_depth_mm'] ].corr()
```

CORRELATION is representing LINEAR relationships!!!!!

I don't like looking at numbers in TABLES! I instead prefer to VISUALIZE things!!!!

Let's visualize the correlation matrix as a HEAT MAP!!!!

This HEAT MAP is given a special name. It's called the **CORRELATION PLOT** or **CORRPLOT** for short!!!!!!

We need to use the `sns.heatmap()` function to make the CORRPLOT. The `sns.heatmap()` function is an AXIS LEVEL FUNCTION.

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            ax=ax)

plt.show()
```

BUT...you should NEVER use the DEFAULT BOUNDS on the COLOR SCALE!!!!

You MUST ALWAYS define the BOUNDS on the color scale to be between -1 and +1 when visualizing CORRPLOTS!!!!!

The bounds are controlled by the `vmin` and `vmax` arguments to `sns.heatmap()`.

AND...you should set the `center` argument which controls the MIDPOINT value between the lower (`vmin`) and upper (`vmax`) color bounds.

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            ax=ax)

plt.show()
```

Diverging color palettes are important to use for CORRPLOTS because we want to focus on behavior AWAY from the MIDPOINT of a correlation coefficient of 0. A diverging color palette lets us distinguish the SIGN (positive vs negative) based on color while still representing the MAGNITUDE by the color intensity/darkness.

The default color palette is `'icefire'`. The palette can be changed with the `cmap` argument.

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            cmap='icefire',
            ax=ax)

plt.show()
```

But there are more diverging color palettes. I like `'coolwarm'`.

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            cmap = 'coolwarm',
            ax=ax)

plt.show()
```

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            cmap='vlag',
            ax=ax)

plt.show()
```

You can ANNOTATE the CORRPLOT via TEXT to make it even easier to display the correlation coefficient!!!

Use the ANNOTATE text to show the EXACT correlation coefficient!

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            cmap='coolwarm',
            annot=True, annot_kws={'size': 25},
            ax=ax)

plt.show()
```

This is for showing 2 specific variables' correlation coefficients.

We can of course change to 2 others...

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.loc[:, ['body_mass_g', 'bill_length_mm']].corr(),
            vmin=-1, vmax=1, center = 0,
            cmap='coolwarm',
            annot=True, annot_kws={'size': 25},
            ax=ax)

plt.show()
```

BUT...correlation plots...are particularly great for showing the relationship between more than just 2 continuous variables...

We can show the correlation coefficient between ALL pairs of continuous variables in one chart!

```python
penguins.corr(numeric_only=True)
```

```python
fig, ax = plt.subplots()

sns.heatmap(data = penguins.corr(numeric_only=True),
            vmin=-1, vmax=1, center=0,
            cmap='coolwarm', cbar=False,
            annot=True, annot_kws={'size': 20},
            ax=ax)

plt.show()
```

## Grouping

Remember we saw that GROUPING BY a categorical CHANGED our opinion about the relationship between two of the continuous variables.

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

We can GROUP BY a categorical variable and apply the `.corr()` method to get the correlation matrix for EACH category!!!

```python
penguins.loc[:, ['flipper_length_mm', 'bill_depth_mm', 'species']].\
groupby(['species']).\
corr()
```

We can visualize the GROUPED correlation plot by iterating over the GROUPS via a for-loop and applying the `sns.heatmap()` function to that group!!!

`sns.heatmap()` is AXIS level function, it will NOT generate facets for us!!!!

It's useful to define helper objects to store the GROUPS we will iterate over.

```python
the_groups = penguins.species.unique().tolist()
```

```python
the_groups
```

```python
corr_per_group = penguins.loc[ :, ['flipper_length_mm', 'bill_depth_mm', 'species']].groupby(['species']).corr()
```

```python
corr_per_group
```

```python
corr_per_group.loc[ 'Adelie' ]
```

```python
corr_per_group.loc[ 'Gentoo' ]
```

```python
corr_per_group.loc[ the_groups[0] ]
```

```python
corr_per_group.loc[ the_groups[1] ]
```

One facet (subplot) per grouping value.

```python
fig, axs = plt.subplots(1, len(the_groups), figsize=(18, 6), sharex=True, sharey=True )

for ix in range(len(the_groups)):
    sns.heatmap( data = corr_per_group.loc[ the_groups[ ix ] ],
                 vmin=-1, vmax=1, center = 0,
                 cmap='coolwarm', cbar=False,
                 annot=True, annot_kws={'size': 20},
                 ax=axs[ix] )
    
    axs[ ix ].set_title('Species: %s' % the_groups[ ix ] )

plt.show()
```

We can create GROUPED correlation plots between more than just 2 continuous variables!!!!

```python
corr_per_group_all = penguins.groupby('species').corr(numeric_only=True)
```

```python
corr_per_group_all
```

```python
fig, axs = plt.subplots(1, len(the_groups), figsize=(18, 6), sharex=True, sharey=True )

for ix in range(len(the_groups)):
    sns.heatmap( data = corr_per_group_all.loc[ the_groups[ ix ] ],
                 vmin=-1, vmax=1, center = 0,
                 cmap='coolwarm', cbar=False,
                 annot=True, annot_kws={'size': 20},
                 ax=axs[ix] )
    
    axs[ ix ].set_title('Species: %s' % the_groups[ ix ] )

plt.show()
```

# Visualizing relationships between continous variables via PAIRS PLOTS

The previous recordings focused mostly on just 2 continuous variables. However, we saw that CORRELATION PLOTS scale to MANY pairs of continuous variables. The PAIRS PLOT is a way to visualize the relationships between MULTIPLE variables in one figure!

## Import Modules

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import seaborn as sns
```

## Read Data

```python
penguins = sns.load_dataset('penguins')
```

```python
penguins.info()
```

## Scatter plot

Visualizes the relationship between 2 continuous variables.

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm')

plt.show()
```

```python
sns.relplot(data = penguins, x='body_mass_g', y='bill_length_mm')

plt.show()
```

```python
penguins.info()
```

We still have not LOOKED at ALL pairs of continuous variables!!!

This might seem like a very tedious task...to make the scatter plot between ALL pairs!

## Pairs plot

The pairs plot is a SHORT CUT!!!! It will create ALL pairs of scatter plots FOR YOU!!!!!

The function is the `sns.pairplot()` function. It is a figure level function because it generates MANY facets!!!

```python
sns.pairplot(data = penguins)

plt.show()
```

Confirm one of the MARGINAL histograms!

```python
sns.displot(data = penguins, x='flipper_length_mm', kind='hist', bins=10)

plt.show()
```

We can GROUP or CONDITION ON  categorical variables within PAIRS PLOTS just like we did with SCATTER PLOTS and TREND PLOTS!!!

We can COLOR by a CATEGORICAL variable via the `hue` argument!!!!!

```python
sns.relplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

```python
sns.lmplot(data = penguins, x='flipper_length_mm', y='bill_depth_mm', hue='species')

plt.show()
```

We can color PAIRS PLOTS via `hue`!!!

```python
sns.pairplot(data = penguins, hue='species')

plt.show()
```

HOWEVER...the above PAIRS PLOT is **WRONG**!!!!!!

Look at the COUNTS for the `species`.

```python
sns.catplot(data = penguins, x='species', kind='count')

plt.show()
```

The DEFAULT pairs plot in Seaborn is WRONG because the CONDITIONAL KDE shown along the MAIN DIAGONAL is FOOLED by the SAMPLE SIZE!!!

We need to remove the sample size effect by setting `common_norm` to FALSE!!!!!!!!

This is the same problem in the CONDITIONAL KDE chart!

The marginal KDE is shown below for one variable.

```python
sns.displot(data = penguins, x='flipper_length_mm', kind='kde')

plt.show()
```

The conditional KDE is shown with the WRONG settings!

```python
sns.displot(data = penguins, x='flipper_length_mm', kind='kde', hue='species')

plt.show()
```

REMOVE the sample size EFFECT with `common_norm=False`.

```python
sns.displot(data = penguins, x='flipper_length_mm', kind='kde', hue='species', common_norm=False)

plt.show()
```

The CONDITIONAL KDE should give you the same conclusion as the VIOLIN PLOT!

```python
sns.catplot(data = penguins, x='species', y='flipper_length_mm', kind='violin', inner='quartile')

plt.show()
```

You must remove the SAMPLE SIZE effect with the PAIRS PLOT just as you did with the CONDITIONIAL KDE!!!!

```python
sns.pairplot(data = penguins, hue='species', 
             diag_kws={'common_norm': False})

plt.show()
```

The pairs plot works best when you have less than 10 variables.

You can get away with up to 15 to 20...but anything more than 20 is INFORMATION OVERLOAD!!!!!!

Because 20 variables...produces a PAIRS PLOT with a 20 X 20 plot matrix!!!!!!!
