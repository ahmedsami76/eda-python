# Introduction to Seaborn

In previous lessons, we used **pandas** to work with tabular data and **Matplotlib** to create plots.

In this lesson, we will learn **Seaborn**, a visualization library built on top of Matplotlib.

Seaborn is especially useful when your data is stored in a pandas DataFrame and you want to quickly explore relationships, comparisons, and distributions.

By the end of this notebook, you should understand:

1. Why Seaborn is useful if you already know Matplotlib.
2. How Seaborn works with pandas DataFrames.
3. How to make simple scatter plots, histograms, and categorical plots.
4. How `hue` helps compare groups.
5. The main differences between Matplotlib and Seaborn.
6. The difference between axes-level and figure-level Seaborn functions.
7. When to use each style of Seaborn function.

# Part 1: Why Learn Seaborn?

Matplotlib is the foundation of Python visualization. It is powerful and flexible, but it can require a lot of manual code.

Seaborn is built on top of Matplotlib. It does **not** replace Matplotlib. Instead, it makes common statistical plots easier to create, especially when working with pandas DataFrames.

A simple way to think about the relationship is:

| Library | Main Role |
|---|---|
| **Matplotlib** | Low-level plotting library. Very flexible, but more manual. |
| **Pandas plotting** | Quick plots directly from DataFrames, built on Matplotlib. |
| **Seaborn** | Higher-level statistical visualization library designed for DataFrames. |

Seaborn is useful because:

1. It works naturally with pandas DataFrames.
2. It has attractive default styles.
3. It can automatically group data using arguments like `hue`.
4. It can create statistical plots with less manual code.
5. It still allows Matplotlib customization when you need it.

## Importing the Libraries

We usually import Seaborn using the alias `sns`.

````python
# If needed, install seaborn first.
# In most notebook environments, this only needs to be done once.

# %pip install seaborn
````

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Optional: set a clean default style for all Seaborn plots
sns.set_theme(style="whitegrid")
````

## Loading an Example Dataset

Seaborn includes a few small example datasets. We will use the `penguins` dataset.

Each row represents one penguin. Each column contains information about that penguin.

````python
penguins = sns.load_dataset("penguins")

penguins.head()
````

````python
penguins.info()
````

Some rows have missing values. For the first examples, we will remove rows with missing values in the columns we plan to use.

This keeps the beginner examples simple.

````python
penguins_clean = penguins.dropna(
    subset=["species", "island", "bill_length_mm", "flipper_length_mm", "body_mass_g", "sex"]
)

penguins_clean.head()
````

# Part 2: The Basic Seaborn Pattern

Most beginner Seaborn plots follow this pattern:

```python
sns.plot_function(data=df, x="column_name", y="column_name")
```

This is one of the biggest differences between Seaborn and basic Matplotlib.

With Matplotlib, we often pass the actual Series or arrays:

```python
ax.scatter(df["column1"], df["column2"])
```

With Seaborn, we usually pass the whole DataFrame to `data`, then refer to columns by name:

```python
sns.scatterplot(data=df, x="column1", y="column2")
```

This style becomes very powerful when we start grouping, coloring, and faceting data.

# Part 3: Matplotlib vs. Seaborn

Let’s compare Matplotlib and Seaborn using the same question:

> Is there a relationship between penguin bill length and flipper length?

First, we will make a basic scatter plot with Matplotlib.

````python
fig, ax = plt.subplots(figsize=(7, 5))

ax.scatter(
    penguins_clean["bill_length_mm"],
    penguins_clean["flipper_length_mm"]
)

ax.set_title("Matplotlib Scatter Plot")
ax.set_xlabel("Bill Length (mm)")
ax.set_ylabel("Flipper Length (mm)")

plt.show()
````

Now let’s make the same plot with Seaborn.

Notice the syntax:

```python
sns.scatterplot(data=penguins_clean, x="bill_length_mm", y="flipper_length_mm")
```

We pass the full DataFrame using `data`, then refer to the columns by name.

````python
sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm"
)

plt.title("Seaborn Scatter Plot")
plt.show()
````

The Seaborn version is not dramatically shorter for this simple plot.

The real advantage appears when we want to group the data.

For example, suppose we want each penguin species to have a different color.

In Matplotlib, we need to manually filter the data by species.

````python
fig, ax = plt.subplots(figsize=(7, 5))

for species in penguins_clean["species"].unique():
    subset = penguins_clean[penguins_clean["species"] == species]
    ax.scatter(
        subset["bill_length_mm"],
        subset["flipper_length_mm"],
        label=species
    )

ax.set_title("Matplotlib: Manual Grouping by Species")
ax.set_xlabel("Bill Length (mm)")
ax.set_ylabel("Flipper Length (mm)")
ax.legend(title="Species")

plt.show()
````

In Seaborn, we can use the `hue` argument.

`hue` means:

> Use color to represent another variable.

Here, we use color to represent the penguin species.

````python
sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species"
)

plt.title("Seaborn: Grouping with hue")
plt.show()
````

This is one of the biggest advantages of Seaborn.

Instead of manually splitting the DataFrame, we tell Seaborn:

- Use this DataFrame.
- Put this column on the x-axis.
- Put this column on the y-axis.
- Use color to separate this group variable.

That is the basic Seaborn workflow.

## Main Differences Between Matplotlib and Seaborn

| Topic | Matplotlib | Seaborn |
|---|---|---|
| Main purpose | General-purpose plotting | Statistical data visualization |
| Works with DataFrames? | Yes, but more manual | Yes, very naturally |
| Syntax style | Often uses arrays or Series directly | Often uses `data=df`, `x="column"`, `y="column"` |
| Grouping | Usually manual | Built in using `hue`, `style`, `size`, `row`, and `col` |
| Default appearance | Basic, customizable | More polished by default |
| Statistical plots | More manual | Many built-in statistical plots |
| Customization | Maximum control | Good defaults, still customizable through Matplotlib |

Important point:

**Seaborn does not replace Matplotlib.**

Seaborn creates Matplotlib figures and axes behind the scenes. You can still use Matplotlib commands like:

```python
plt.title()
plt.xlabel()
plt.ylabel()
plt.show()
```

or, when working with Matplotlib axes:

```python
ax.set_title()
ax.set_xlabel()
ax.set_ylabel()
```

# Part 4: A Few Basic Seaborn Plots

Before learning Seaborn architecture, let’s see a few common plot types.

We will look at:

1. Scatter plots for relationships.
2. Histograms for distributions.
3. Box plots for comparing categories.
4. Bar plots for automatic statistical summaries.

## Scatter Plot: Relationship Between Two Numeric Variables

Use a scatter plot when both variables are numeric.

Here, both `bill_length_mm` and `flipper_length_mm` are numeric variables.

````python
sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species"
)

plt.title("Bill Length vs. Flipper Length by Species")
plt.show()
````

## Histogram: Distribution of One Numeric Variable

Use a histogram to see the distribution of one numeric variable.

Here, we look at the distribution of penguin body mass.

````python
sns.histplot(
    data=penguins_clean,
    x="body_mass_g"
)

plt.title("Distribution of Penguin Body Mass")
plt.show()
````

We can also use `hue` with a histogram to compare distributions by group.

````python
sns.histplot(
    data=penguins_clean,
    x="body_mass_g",
    hue="species"
)

plt.title("Body Mass Distribution by Species")
plt.show()
````

## Box Plot: Comparing a Numeric Variable Across Categories

Use a box plot when you want to compare a numeric variable across groups.

Here, we compare body mass across species.

- `species` is categorical.
- `body_mass_g` is numeric.

````python
sns.boxplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g"
)

plt.title("Body Mass by Species")
plt.show()
````

## Bar Plot: Automatic Statistical Summary

A Seaborn `barplot()` is different from a basic Matplotlib bar chart.

In Seaborn, `barplot()` usually starts with raw data and automatically calculates a summary statistic.

By default, Seaborn calculates the **mean** of the numeric variable for each category.

Here, Seaborn calculates the average body mass for each penguin species.

````python
sns.barplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    errorbar=("ci", 95)
)

plt.title("Average Body Mass by Species")
plt.show()
````

We can also change the summary statistic using the `estimator` argument.

For example, we can show the median instead of the mean.

````python
sns.barplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    estimator=np.median,
    errorbar=None
)

plt.title("Median Body Mass by Species")
plt.show()
````

Important note:

`barplot()` shows a summary, not every individual observation.

If you want to show the raw data points, you can use `stripplot()` or `swarmplot()`.

````python
sns.stripplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    hue="species",
    jitter=True,
    alpha=0.6,
    legend=False
)

plt.title("Raw Body Mass Values by Species")
plt.show()
````

We can even combine two Seaborn plots on the same axes.

For example, the next plot shows:

- Raw data points using `stripplot()`.
- The mean for each group using `pointplot()`.

This is a good example of why axes-level functions are useful: they can be layered on top of each other.

````python
fig, ax = plt.subplots(figsize=(7, 5))

sns.stripplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    color="gray",
    jitter=True,
    alpha=0.4,
    ax=ax
)

sns.pointplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    estimator=np.mean,
    errorbar=None,
    color="red",
    markers="D",
    linestyles="",
    ax=ax
)

ax.set_title("Raw Data Points with Group Means")
ax.set_ylabel("Body Mass (g)")

plt.show()
````

At this point, we have already seen the most important beginner pattern:

```python
sns.plot_function(data=df, x="column", y="column", hue="group")
```

**axes-level vs. figure-level functions.**

# Part 5: Axes-Level vs. Figure-Level Functions

Seaborn has two main types of functions:

1. **Axes-level functions**
2. **Figure-level functions**

This sounds complicated at first, but the idea is simple.

## Matplotlib Reminder: Figure vs. Axes

In Matplotlib:

- A **Figure** is the whole canvas.
- An **Axes** is one individual plot inside the figure.

For example:

```python
fig, ax = plt.subplots()
```

Here:

- `fig` is the full figure.
- `ax` is one plot area.

If we create multiple subplots:

```python
fig, axes = plt.subplots(1, 2)
```

Then we have:

- One figure.
- Two axes.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].set_title("Axes 1")
axes[0].text(0.5, 0.5, "First plot area", ha="center", va="center", fontsize=12)

axes[1].set_title("Axes 2")
axes[1].text(0.5, 0.5, "Second plot area", ha="center", va="center", fontsize=12)

fig.suptitle("One Figure with Two Axes", fontsize=14, fontweight="bold")

plt.tight_layout()
plt.show()
````

## Axes-Level Functions

Axes-level functions draw on one existing plot area.

Common axes-level functions include:

- `sns.scatterplot()`
- `sns.lineplot()`
- `sns.histplot()`
- `sns.kdeplot()`
- `sns.boxplot()`
- `sns.violinplot()`
- `sns.barplot()`
- `sns.stripplot()`

These functions return a Matplotlib `Axes` object.

Use axes-level functions when:

- You want one plot.
- You want to customize the plot using Matplotlib.
- You want to place several plots in your own subplot layout.
- You want to overlay multiple plots on the same axes.

````python
fig, ax = plt.subplots(figsize=(7, 5))

sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    ax=ax
)

ax.set_title("Axes-Level Function: sns.scatterplot()")
ax.set_xlabel("Bill Length (mm)")
ax.set_ylabel("Flipper Length (mm)")

plt.show()
````

In this example, we created the figure ourselves using Matplotlib:

```python
fig, ax = plt.subplots()
```

Then we told Seaborn exactly where to draw:

```python
ax=ax
```

That is the axes-level approach.

## Figure-Level Functions

Figure-level functions create the whole figure for you.

Common figure-level functions include:

- `sns.relplot()`
- `sns.catplot()`
- `sns.displot()`
- `sns.lmplot()`
- `sns.jointplot()`
- `sns.pairplot()`

These functions are especially useful when you want Seaborn to automatically create multiple subplots.

The most important arguments are:

- `col`: split plots across columns.
- `row`: split plots across rows.

This is called **faceting**.

````python
g = sns.relplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    height=5,
    aspect=1.3
)

g.fig.suptitle("Figure-Level Function: sns.relplot()", y=1.03)

plt.show()
````

This looks similar to `sns.scatterplot()`, but there is an important difference.

`sns.relplot()` creates the figure for us.

Now let’s see why figure-level functions are useful.

````python
g = sns.relplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    col="island",
    height=4,
    aspect=1
)

g.fig.suptitle("Faceting with col='island'", y=1.05)

plt.show()
````

With just one extra argument:

```python
col="island"
```

Seaborn created one subplot for each island.

Doing this manually in Matplotlib would require more code.

## How Axes-Level and Figure-Level Functions Are Related

A useful way to remember the difference:

| If you want... | Use... | Example |
|---|---|---|
| One plot on one axes | Axes-level | `sns.scatterplot()` |
| Your own subplot layout | Axes-level | `sns.boxplot(ax=axes[0])` |
| Automatic subplots by category | Figure-level | `sns.relplot(col="island")` |
| Quick faceted exploration | Figure-level | `sns.catplot(col="species")` |

Many figure-level functions are wrappers around axes-level functions.

| Figure-Level Function | Often Uses Axes-Level Function |
|---|---|
| `relplot(kind="scatter")` | `scatterplot()` |
| `relplot(kind="line")` | `lineplot()` |
| `catplot(kind="box")` | `boxplot()` |
| `catplot(kind="violin")` | `violinplot()` |
| `displot(kind="hist")` | `histplot()` |
| `displot(kind="kde")` | `kdeplot()` |

So you do not need to think of these as completely separate systems.

The simple rule is:

> Use axes-level functions when you want control.  
> Use figure-level functions when you want Seaborn to create the full layout for you.

# Part 6: Axes-Level Functions for Custom Layouts

Here we create our own 1-by-2 subplot layout using Matplotlib, then draw Seaborn plots into each axes.

This is a good use case for axes-level functions because we want to design the subplot layout ourselves.

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    ax=axes[0]
)

axes[0].set_title("Scatter Plot")

sns.boxplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    ax=axes[1]
)

axes[1].set_title("Box Plot")

plt.tight_layout()
plt.show()
````

We can also create a larger grid and mix different Seaborn plot types.

````python
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Scatter plot
sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    ax=axes[0, 0]
)
axes[0, 0].set_title("Scatter: Bill Length vs. Flipper Length")

# Plot 2: Histogram
sns.histplot(
    data=penguins_clean,
    x="bill_length_mm",
    kde=True,
    ax=axes[0, 1]
)
axes[0, 1].set_title("Distribution: Bill Length")

# Plot 3: Box plot
sns.boxplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    ax=axes[1, 0]
)
axes[1, 0].set_title("Box Plot: Body Mass by Species")

# Plot 4: Violin plot
sns.violinplot(
    data=penguins_clean,
    x="species",
    y="flipper_length_mm",
    ax=axes[1, 1]
)
axes[1, 1].set_title("Violin Plot: Flipper Length by Species")

plt.tight_layout()
plt.show()
````

# Part 7: Figure-Level Functions for Automatic Faceting

Now we let Seaborn create the subplots automatically.

This is a good use case for figure-level functions because we want separate plots for each category.

````python
g = sns.catplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    col="island",
    kind="box",
    height=4,
    aspect=1
)

g.fig.suptitle("Body Mass by Species, Split by Island", y=1.05)

plt.show()
````

The argument:

```python
col="island"
```

automatically created one column of plots for each island.

The argument:

```python
kind="box"
```

told `catplot()` to create box plots.

We can change the type of categorical plot by changing `kind`.

For example, `kind="violin"` creates violin plots.

A violin plot shows the shape of the distribution for each category. It is like a box plot with more information about the distribution shape.

````python
g = sns.catplot(
    data=penguins_clean,
    x="species",
    y="flipper_length_mm",
    kind="violin",
    inner="quartile",
    height=5,
    aspect=1.2
)

g.fig.suptitle("Flipper Length Distribution by Species", y=1.03)

plt.show()
````

# Part 8: Aesthetic Mappings

In Seaborn, an aesthetic mapping means using a visual property to represent a variable.

The most common aesthetic mappings are:

| Argument | Meaning |
|---|---|
| `x` | Put a variable on the x-axis |
| `y` | Put a variable on the y-axis |
| `hue` | Use color to represent a variable |
| `size` | Use marker size to represent a variable |
| `style` | Use marker shape or line style to represent a variable |

We already used `hue`. Now let’s look at a few more examples.

## Using `hue`

`hue` maps a variable to color.

Here, color represents species.

````python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    ax=axes[0]
)
axes[0].set_title("Without hue")

sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    ax=axes[1]
)
axes[1].set_title("With hue='species'")

plt.tight_layout()
plt.show()
````

## Using `size` and `style`

We can also encode more variables using `size` and `style`.

However, be careful: adding too many visual encodings can make a plot hard to read.

For beginners, the most important aesthetic is usually `hue`.

````python
sns.scatterplot(
    data=penguins_clean,
    x="bill_length_mm",
    y="flipper_length_mm",
    hue="species",
    size="body_mass_g",
    style="sex",
    sizes=(40, 200),
    alpha=0.7
)

plt.title("Using hue, size, and style")
plt.show()
````

This plot shows several variables at once:

- `x`: bill length
- `y`: flipper length
- `hue`: species
- `size`: body mass
- `style`: sex

This can be useful, but it can also become visually crowded.

A good rule for beginners:

> Start simple. Add `hue` first. Only add `size` or `style` if it makes the plot easier to understand.

# Part 9: Statistical Estimation in Seaborn

One of Seaborn’s strengths is that it can calculate statistics from raw data.

For example, `barplot()` can calculate the mean or median for each category.

Let’s compare a few options.

````python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Default: mean with 95% confidence interval
sns.barplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    errorbar=("ci", 95),
    ax=axes[0]
)
axes[0].set_title("Mean with 95% CI")

# Median without error bars
sns.barplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    estimator=np.median,
    errorbar=None,
    ax=axes[1]
)
axes[1].set_title("Median")

# Raw observations
sns.stripplot(
    data=penguins_clean,
    x="species",
    y="body_mass_g",
    hue="species",
    jitter=True,
    alpha=0.6,
    legend=False,
    ax=axes[2]
)
axes[2].set_title("Raw Observations")

plt.tight_layout()
plt.show()
````

The key idea is:

> Seaborn can work directly with raw data and calculate useful summaries for you.

But you should always know whether your plot is showing:

- Raw observations.
- Summary statistics.
- Both.

# Part 10: Pair Plots for Quick Exploration

Now that we have learned the basics, we can introduce `pairplot()`.

A pair plot is a quick way to explore relationships between several numeric variables at once.

It creates a grid of plots:

- Each off-diagonal plot shows the relationship between two numeric variables.
- Each diagonal plot shows the distribution of one numeric variable.

Pair plots are useful for exploratory data analysis, but they can become hard to read if you include too many variables.

````python
sns.pairplot(
    data=penguins_clean,
    vars=["bill_length_mm", "bill_depth_mm", "flipper_length_mm", "body_mass_g"],
    hue="species"
)

plt.show()
````

Pair plots work best with a small number of numeric variables.

Why?

If you include 4 numeric variables, Seaborn creates a 4 by 4 grid.

If you include 10 numeric variables, Seaborn creates a 10 by 10 grid.

That is already 100 small plots.

A good rule of thumb:

> Use `pairplot()` with a small number of carefully selected variables.

# Part 11: Decision Guide

Here is a simple guide for choosing between Seaborn approaches.

| Situation | Recommended Approach | Example |
|---|---|---|
| I want one simple plot | Axes-level function | `sns.scatterplot()` |
| I want to customize titles, labels, or layout manually | Axes-level function with Matplotlib | `fig, ax = plt.subplots()` then `sns.scatterplot(ax=ax)` |
| I want multiple subplots based on a category | Figure-level function | `sns.relplot(col="island")` |
| I want a categorical plot with automatic faceting | Figure-level function | `sns.catplot(col="island", kind="box")` |
| I want quick exploratory plots | Figure-level function | `sns.pairplot()` or `sns.relplot()` |
| I want full control over a custom subplot grid | Axes-level functions | `plt.subplots()` plus multiple Seaborn calls |

Another way to think about it:

```text
START
  |
  |-- Do you need automatic subplots by category?
  |       |
  |       |-- YES --> Use a figure-level function
  |       |            Examples: relplot(), catplot(), displot()
  |
  |       |-- NO --> Continue
  |
  |-- Do you need a custom subplot layout or want to overlay plots?
          |
          |-- YES --> Use axes-level functions with plt.subplots()
          |            Examples: scatterplot(), boxplot(), histplot()
          |
          |-- NO --> Use either style
                       For beginners, start with the simpler one
```
