# Part 1: Why Seaborn?

## The Matplotlib vs. Seaborn Relationship

You might be wondering: *If Pandas uses Matplotlib, why do I need to learn Seaborn?*

Think of **Matplotlib** as the 'Engine' of Python visualization. It is powerful and gives you total control over every single pixel, but it is low-level and verbose. To make a simple chart look professional in Matplotlib, you often end up writing 10-15 lines of code for labels, legends, and styling.

**Seaborn** is the 'Car Body' built on top of that engine. It is designed specifically for statistical data exploration. Here is why we use it:

1. **Pandas Integration:** Seaborn is built to work *directly* with DataFrames. You pass the whole DataFrame in, and tell it which column is `x` and which is `y`.
2. **Statistical Defaults:** It comes with beautiful, readable color palettes and styles by default. No more tweaking font sizes and background colors for 30 minutes.
3. **Complex Logic in One Line:** Seaborn handles aggregation (calculating means/confidence intervals) and categorical splits (using the `hue` argument) automatically.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Seaborn is standardly imported as sns
import seaborn as sns
````

## Comparison: Matplotlib vs. Seaborn
Let's see the difference in syntax using the classic 'penguins' dataset.

````python
# Load example dataset
penguins = sns.load_dataset('penguins')

# 1. THE MATPLOTLIB WAY: Manual work
# We have to filter by species manually and call plot() multiple times
fig, ax = plt.subplots()
for species in penguins['species'].unique():
    subset = penguins[penguins['species'] == species]
    ax.scatter(subset['bill_length_mm'], subset['flipper_length_mm'], label=species)
ax.legend()
ax.set_title('Matplotlib: Manual Grouping')
plt.show()
````

````python

# 2. THE SEABORN WAY: Expressive and automatic
# We just pass the DataFrame and the 'hue' argument handles the split!
sns.scatterplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', hue='species')
plt.title('Seaborn: Automatic Grouping')
plt.show()
````

# Part 2: Categorical Plotting

Seaborn truly shines when comparing categorical variables.

````python
# Create a violin plot to see distribution of flipper length by species
# kind='violin' handles the density estimation automatically
sns.catplot(data=penguins, x='species', y='flipper_length_mm', kind='violin', inner='quartile')

plt.show()
````

# Part 3: Seaborn Architecture & Programming Paradigms

## Understanding Seaborn's Structure

Seaborn is organized around a simple but powerful concept: it provides functions to create statistical graphics. However, these functions fall into two categories:

1. **Axes-Level Functions** - Work with Matplotlib's `Axes` objects
2. **Figure-Level Functions** - Work with Matplotlib's `Figure` objects and handle subplot creation automatically

Let's explore what this means and why it matters.

---

## Axes-Level vs Figure-Level: The Core Distinction

### Axes-Level Functions
- **What they do:** Plot onto a single `Axes` object (a single subplot)
- **Returns:** The `Axes` object itself
- **You control:** Subplot creation, layout, and multiple plots on one figure
- **When to use:** When you need fine-grained control or want to combine plots

**Common axes-level functions:**
- `scatterplot()`, `lineplot()`, `barplot()`, `boxplot()`, `violinplot()`, `histplot()`, `kdeplot()`

### Figure-Level Functions
- **What they do:** Automatically create the figure and handle multiple subplots
- **Returns:** A `FacetGrid` object (which contains the figure)
- **You control:** The overall figure and faceting (splitting into subplots by categories)
- **When to use:** When you want quick, automatic subplot creation

**Common figure-level functions:**
- `relplot()`, `catplot()`, `displot()`, `lmplot()`, `jointplot()`

---


## Example 1: Axes-Level Functions (You Control the Figure)

With **axes-level functions**, you explicitly create the figure and subplots using Matplotlib, then pass the `Axes` object to Seaborn.


````python
# AXES-LEVEL APPROACH
# Step 1: Create the figure and axes using Matplotlib
fig, ax = plt.subplots(figsize=(8, 5))

# Step 2: Pass the axes to a Seaborn axes-level function
sns.scatterplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', 
                hue='species', ax=ax)

# Step 3: YOU control the final touches
ax.set_title('Axes-Level: Manual Control', fontsize=14, fontweight='bold')
ax.set_xlabel('Bill Length (mm)', fontsize=12)
ax.set_ylabel('Flipper Length (mm)', fontsize=12)

plt.tight_layout()
plt.show()

print("Key Point: You explicitly created the figure and axes!")
````

## Example 2: Figure-Level Functions (Seaborn Controls the Figure)

With **figure-level functions**, Seaborn automatically creates the figure and handles subplot management. This is especially powerful for faceting (creating multiple subplots based on categorical variables).


````python
# FIGURE-LEVEL APPROACH (Simple Version)
# Seaborn creates the figure automatically!
sns.relplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', 
            hue='species', height=5, aspect=1.5)

plt.suptitle('Figure-Level: Automatic Figure Management', fontsize=14, fontweight='bold', y=1.00)
plt.show()

print("Key Point: Seaborn created the entire figure for you!")
````

## Example 3: Faceting with Figure-Level Functions

The real power of figure-level functions is **faceting** — creating multiple subplots automatically based on a categorical variable. The `col` parameter splits your data across columns, and `row` splits across rows.


````python
# FIGURE-LEVEL WITH FACETING
# Create separate subplots for each species automatically!
g = sns.relplot(data=penguins, 
                x='bill_length_mm', y='flipper_length_mm',
                col='species',  # Split by species across columns
                height=4, aspect=1)

g.set_titles('{col_name} Species')
plt.show()

print("Key Point: Faceting creates multiple subplots automatically based on a category!")
print("Notice: One line of code with col='species' replaces multiple matplotlib subplot calls.")
````

## Quick Comparison: When to Use Each Approach

| Scenario | Use | Example Function |
|----------|-----|-----------------|
| **Quick exploratory plot with simple styling** | Figure-Level | `relplot()`, `catplot()` |
| **Need faceting (split by category)** | Figure-Level | `relplot(col='species')`, `catplot()` |
| **Combining multiple plots on one figure** | Axes-Level | `scatterplot(ax=ax1)` + `scatterplot(ax=ax2)` |
| **Fine-tuned control over subplot layout** | Axes-Level | Custom `subplots()` with specific positioning |
| **Overlaying multiple plot types** | Axes-Level | `scatterplot()` + `lineplot()` on same axes |
| **Interactive/conditional styling per subplot** | Axes-Level | Loop through groups, customize each subplot |

---

## Example 4: Combining Multiple Axes-Level Plots

One advantage of axes-level functions is that you can layer multiple plots on the same axes.


````python
# MULTIPLE AXES-LEVEL PLOTS ON ONE FIGURE
fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# Plot 1: Scatter plot
sns.scatterplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', 
                hue='species', ax=axes[0, 0])
axes[0, 0].set_title('Scatterplot: Bill Length vs Flipper Length')

# Plot 2: Distribution (histogram + KDE)
sns.histplot(data=penguins, x='bill_length_mm', kde=True, ax=axes[0, 1])
axes[0, 1].set_title('Distribution: Bill Length')

# Plot 3: Categorical boxplot
sns.boxplot(data=penguins, x='species', y='body_mass_g', ax=axes[1, 0])
axes[1, 0].set_title('Boxplot: Body Mass by Species')

# Plot 4: Categorical violin plot
sns.violinplot(data=penguins, x='species', y='flipper_length_mm', ax=axes[1, 1])
axes[1, 1].set_title('Violinplot: Flipper Length by Species')

plt.tight_layout()
plt.show()

print("Key Point: With axes-level functions, you control the entire grid layout and can mix different plot types!")
````

## Part 4: Core Seaborn Programming Techniques

### Technique 1: Aesthetic Mapping with `hue`

The `hue` parameter maps a categorical variable to color. This is one of Seaborn's most powerful features.


````python
# The power of 'hue': Splits a categorical variable by color
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# WITHOUT hue: Hard to compare groups
sns.scatterplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', ax=axes[0])
axes[0].set_title('Without hue: All points look the same')

# WITH hue: Groups are visually separated
sns.scatterplot(data=penguins, x='bill_length_mm', y='flipper_length_mm', 
                hue='species', ax=axes[1], s=100)
axes[1].set_title('With hue="species": Each group has its own color')

plt.tight_layout()
plt.show()
````

### Technique 2: Multi-Dimensional Aesthetics with `size` and `style`

You can encode up to **4 variables** at once: `x`, `y`, `hue`, and `size` (or `style`).


````python
# Add a 'size' aesthetic to encode a 4th variable
sns.scatterplot(data=penguins, 
                x='bill_length_mm', 
                y='flipper_length_mm',
                hue='species',           # Color encodes species
                size='body_mass_g',      # Size encodes body mass
                sizes=(50, 200),         # Size range
                alpha=0.6)

plt.title('Multi-Dimensional Aesthetics: 4 Variables Visualized\n(x, y, hue=color, size=mass)', 
          fontsize=12, fontweight='bold')
plt.show()

print("4 variables: bill_length (x), flipper_length (y), species (color), body_mass (size)")
````

### Technique 3: Automatic Aggregation with `estimator`

Categorical plot functions like `barplot()` and `pointplot()` automatically compute statistics (mean, median, etc.) from raw data.


````python
# Compare different aggregation approaches
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Default: Mean with 95% confidence interval
sns.barplot(data=penguins, x='species', y='body_mass_g', ax=axes[0])
axes[0].set_title('Barplot (Default): Mean ± 95% CI')

# You can change the estimator
sns.barplot(data=penguins, x='species', y='body_mass_g', 
            estimator='median', ax=axes[1])
axes[1].set_title('Barplot: Median')

# Show raw points + mean
sns.pointplot(data=penguins, x='species', y='body_mass_g', 
              ax=axes[2], join=False)
axes[2].set_title('Pointplot: Mean with individual points')

plt.tight_layout()
plt.show()

print("Key: Seaborn computes statistics automatically! You don't have to manually aggregate.")
````

## Summary: The Seaborn Toolkit

| Aspect | What It Does | Example |
|--------|-------------|---------|
| **Aesthetics Mapping** | Encode multiple variables as visual properties | `hue='species'`, `size='mass'` |
| **Axes-Level Functions** | Control individual subplots | `scatterplot()`, `barplot()` |
| **Figure-Level Functions** | Auto-create figure & enable faceting | `relplot()`, `catplot()` |
| **Faceting** | Split data into subplots by category | `col='species'`, `row='island'` |
| **Statistical Estimation** | Auto-aggregate & compute statistics | `estimator='median'`, confidence intervals |
| **Color Palettes** | Professional color schemes | `palette='husl'`, `palette='Set2'` |

---

## Decision Tree: Choosing Your Approach

```
START
  ↓
Do you need FACETING (subplots by category)?
  ├─ YES → Use FIGURE-LEVEL (relplot, catplot, displot)
  └─ NO  → Continue...
           ↓
        Do you need to COMBINE multiple plots?
           ├─ YES → Use AXES-LEVEL (scatterplot, barplot) + plt.subplots()
           └─ NO  → Use either (Figure-level is simpler!)
```

**Pro Tips:**
- Start with figure-level for quick exploration; switch to axes-level for customization
- Always pass your DataFrame first: `sns.plot(data=df, ...)`
- Use `hue` to compare groups instantly
- Remember: Seaborn handles the math; you focus on the story
